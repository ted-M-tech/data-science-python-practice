# 1 INNER JOIN
Select employee names, salaries, and their respective department names. Only include employees assigned to a department.


```python
query = """
SELECT first_name, last_name, salary, departments.department_name
FROM employees
JOIN departments ON employees.department_id = departments.department_id;
"""

df = pd.read_sql_query(query, conn)
df
```

# 2 LEFT JOIN
List all employees along with their department names. Show 'Unassigned' for employees without a department

```python
query = """
SELECT first_name, last_name, 
CASE WHEN departments.department_name is null THEN 'Unaasined'
ELSE departments.department_name END
FROM employees
LEFT JOIN departments ON employees.department_id = departments.department_id;
"""

df = pd.read_sql_query(query, conn)
df
```

# 3 RIGHT JOIN
Display each department name along with employee names. Include departments that have no employees.

```python
query = """
SELECT
    departments.department_name, first_name, last_name
FROM
    employees
RIGHT JOIN
    departments 
ON
    employees.department_id = departments.department_id;
"""

df = pd.read_sql_query(query, conn)
df
```

# 4 FULL OUTER JOIN
Select all employees and all departments, including employees without departments and departments without employees.

```python
query = """
SELECT
    departments.department_name, first_name, last_name
FROM
    employees
FULL OUTER JOIN
    departments 
ON 
    employees.department_id = departments.department_id;
"""

df = pd.read_sql_query(query, conn)
# pd.set_option('display.max_rows', None)

df
```

# 5 CROSS JOIN
Create a query to show all possible combinations of employee names and department names.

```python
query = """
SELECT 
    d.department_name, e.first_name, e.last_name
FROM 
    employees AS e
CROSS JOIN 
    departments AS d;
"""

df = pd.read_sql_query(query, conn)

df
```

# 6 SELF JOIN
Using a self join, write a query to display each employee along with their manager's name. For employees without a manager, display 'No Manager'.

```python
query = """
SELECT
    e.first_name || ' ' || e.last_name AS employee_name,
    CASE WHEN m.first_name IS NOT NULL THEN m.first_name || ' ' || m.last_name
    ELSE 'No Manager' END AS manager_name
FROM
    employees e  -- Alias 'e' for the Employee
LEFT JOIN
    employees m  -- Alias 'm' for the Manager
ON
    e.manager_id = m.employee_id;
"""

df = pd.read_sql_query(query, conn)
# pd.set_option('display.max_rows', None)

df
```

# 7 INNER JOIN with Condition
Select employees whose salary is above the average salary of their department.

```python
query = """
SELECT
    e.first_name,
    e.last_name,
    e.salary,
    d.department_name
FROM
    employees e
JOIN
    departments d ON e.department_id = d.department_id
WHERE
    e.salary > (
        SELECT AVG(e_sub.salary)
        FROM employees e_sub
        WHERE e_sub.department_id = e.department_id
    );
"""

df = pd.read_sql_query(query, conn)
df
```

# 8 LEFT JOIN with Aggregation
Calculate the average salary for each department. Include departments with no employees, showing a count of 0 and an average salary as 'NULL'.

```python
query = """
SELECT
    d.department_name,
    COUNT(e.employee_id) AS employee_count,
    AVG(e.salary) AS average_salary
FROM
    departments d
LEFT JOIN
    employees e ON d.department_id = e.department_id
GROUP BY
    d.department_name
ORDER BY
    d.department_name;
"""

df = pd.read_sql_query(query, conn)
df
```

# 9 RIGHT JOIN with Aggregation
Count the number of employees in each department, including departments with no employees. Sort by the number of employees in descending order.

```python
query = """
SELECT
    d.department_name,
    COUNT(e.employee_id) AS employee_count
FROM
    employees e
RIGHT JOIN
    departments d ON e.department_id = d.department_id
GROUP BY
    d.department_name
ORDER BY
    employee_count DESC;
"""

df = pd.read_sql_query(query, conn)
df
```

# 10 FULL OUTER JOIN with Advanced Conditions
Write a query to select all employees and department names, including employees without departments and departments without employees. Display a meaningful placeholder for missing data.

```python
query = """
SELECT
    CASE WHEN first_name IS NULL THEN 'No name'
    ELSE first_name || ' ' || last_name END AS employee_name,
    CASE WHEN departments.department_name IS NULL THEN 'No department'
    ELSE departments.department_name END
FROM
    employees
FULL OUTER JOIN
    departments 
ON 
    employees.department_id = departments.department_id;
"""

df = pd.read_sql_query(query, conn)
# pd.set_option('display.max_rows', None)

df
```
