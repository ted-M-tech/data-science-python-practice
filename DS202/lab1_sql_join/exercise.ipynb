# Exercise 1

Select all columns from the movies table.

```python
cur.execute("SELECT * FROM movies;")
rows = pd.DataFrame(cur.fetchall())
print(rows)
```

# Exercise 2

Select only the title and release_year columns from the movies table.

```python
cur.execute("SELECT title, release_year FROM movies;")
rows = pd.DataFrame(cur.fetchall(), columns=['title', 'release_year'])
print(rows)
```
