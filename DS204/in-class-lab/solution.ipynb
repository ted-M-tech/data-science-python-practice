# In-Class Lab: Pearson Correlation Test Using Real World Advertising Data

## Objective
Explore the relationship between advertising spending (TV, Radio, Newspaper) and sales revenue using the Pearson correlation coefficient and visualizations.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr

df = pd.read_csv('Advertising.csv', index_col=0)
df.head()
```

## Part 1: Data Familiarization

### Identify research questions:
- Is there a significant relationship between TV advertising and sales?
- How does radio advertising correlate with sales?
- Does newspaper advertising have a strong correlation with sales?

```python
df.info()
```

```python
# Display descriptive statistics
df.describe()
```

## Part 2: Pearson Correlation Analysis

```python
# Calculate Pearson correlation coefficients
corr_tv, _ = pearsonr(df['TV'], df['Sales'])
corr_radio, _ = pearsonr(df['Radio'], df['Sales'])
corr_newspaper, _ = pearsonr(df['Newspaper'], df['Sales'])

print(f"Correlation between TV and Sales: {corr_tv:.2f}")
print(f"Correlation between Radio and Sales: {corr_radio:.2f}")
print(f"Correlation between Newspaper and Sales: {corr_newspaper:.2f}")
```

## Part 3: Data Visualization
Scatterplots with trendlines to visualize the relationships.

```python
# Set up the figure
fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

# TV vs Sales
sns.regplot(ax=axes[0], x='TV', y='Sales', data=df, line_kws={'color': 'red'})
axes[0].set_title(f'TV vs Sales (r={corr_tv:.2f})')

# Radio vs Sales
sns.regplot(ax=axes[1], x='Radio', y='Sales', data=df, line_kws={'color': 'red'})
axes[1].set_title(f'Radio vs Sales (r={corr_radio:.2f})')

# Newspaper vs Sales
sns.regplot(ax=axes[2], x='Newspaper', y='Sales', data=df, line_kws={'color': 'red'})
axes[2].set_title(f'Newspaper vs Sales (r={corr_newspaper:.2f})')

plt.tight_layout()
plt.show()
```

## Part 4: Reporting and Presentation

**1. TV Advertising and Sales:**
There is a strong positive correlation between TV advertising and Sales. As TV advertising budget increases, sales tend to increase significantly.

**2. Radio Advertising and Sales:**
There is a moderate positive correlation between Radio advertising and Sales. It contributes to sales but less strongly than TV.

**3. Newspaper Advertising and Sales:**
There is a weak positive correlation between Newspaper advertising and Sales. The relationship is not as clear or strong as the other mediums.

**Conclusion:**
TV advertising appears to be the most effective medium for driving sales in this dataset, followed by Radio. Newspaper advertising has the least impact.
