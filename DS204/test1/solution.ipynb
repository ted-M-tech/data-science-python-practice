```python
import math
from scipy import stats

# Function to perform Z-Test and print results
def perform_z_test(question_num, claim_desc, mu_claim, x_bar, sigma, n, test_type):
    # 1. State Hypotheses based on test type
    if test_type == 'two-tailed':
        h0 = f"mu = {mu_claim}"
        h1 = f"mu != {mu_claim}"
    elif test_type == 'left-tailed': # Used when testing if value is "less than" or rejecting "at least"
        h0 = f"mu >= {mu_claim}"
        h1 = f"mu < {mu_claim}"
    elif test_type == 'right-tailed': # Used when testing if value is "greater than" or rejecting "at most"
        h0 = f"mu <= {mu_claim}"
        h1 = f"mu > {mu_claim}"
    
    # 2. Calculate Z-score
    # Standard Error = sigma / sqrt(n)
    se = sigma / math.sqrt(n)
    z_score = (x_bar - mu_claim) / se
    
    # 3. Calculate p-value
    if test_type == 'two-tailed':
        p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))
    elif test_type == 'left-tailed':
        p_value = stats.norm.cdf(z_score)
    elif test_type == 'right-tailed':
        p_value = 1 - stats.norm.cdf(z_score)
        
    # 4. Conclusion
    alpha = 0.05
    if p_value < alpha:
        conclusion = "Reject Null Hypothesis (H0)"
    else:
        conclusion = "Fail to reject Null Hypothesis (H0)"
        
    # Print Output
    print(f"--- Question {question_num}: {claim_desc} ---")
    print(f"Hypotheses: H0: {h0} | H1: {h1}")
    print(f"Z-score:    {z_score:.4f}")
    print(f"p-value:    {p_value:.4f}")
    print(f"Conclusion: {conclusion}")
```

# Question 1: Average Time on a Website
- Scenario: A website owner claims that the average time a visitor spends on their
site is 10 minutes. 
- A sample of 100 visitors shows an average time of 9.8 minutes with a population standard deviation of 1.5 minutes.
- Task: Test the website owner's claim at a 5% significance level.

```python
# Q1: Claim mean is 10 min. Sample 9.8.
perform_z_test(1, "Website Time", 10, 9.8, 1.5, 100, 'two-tailed')
```
